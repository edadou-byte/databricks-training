# NB02 - Création de table de streaming avec Auto Loader

## Qu'est-ce qu'une table de streaming ?

Une table de streaming est une table qui reçoit en continu des données provenant d'une source (comme des fichiers, des messages Kafka, etc.). Elle permet de traiter, transformer et analyser les données en temps réel à mesure qu'elles arrivent, sans attendre que l'ensemble des données soit disponible.

## Qu'est-ce que l'Auto Loader ?

L'Auto Loader est une fonctionnalité de Databricks qui permet de charger automatiquement et efficacement des données en continu depuis des fichiers nouveaux ou modifiés dans un dossier (par exemple, sur un stockage cloud). Il détecte automatiquement les nouveaux fichiers, les lit de façon incrémentale et les ingère dans une table de streaming, simplifiant ainsi la gestion et la scalabilité des pipelines de données.

## Dans quels cas utiliser les tables de streaming ?

Les tables de streaming sont particulièrement utiles dans les situations suivantes :

- **Traitement en temps réel** : Lorsque vous devez analyser ou transformer des données dès leur arrivée (par exemple, détection de fraudes, monitoring, alertes).
- **Ingestion continue de données** : Pour intégrer en continu des données provenant de sources comme des logs, des capteurs IoT, ou des flux d'événements.
- **Mise à jour fréquente des données** : Lorsque les données changent fréquemment et que vous souhaitez que vos analyses ou tableaux de bord reflètent les dernières informations.
- **Réduction de la latence** : Pour réduire le délai entre l'arrivée des données et leur disponibilité pour l'analyse ou la prise de décision.
- **Automatisation des pipelines de données** : Pour automatiser le traitement et l'intégration de nouveaux fichiers ou événements sans intervention manuelle.

En résumé, utilisez les tables de streaming dès que vous avez besoin de traiter des données en continu et en temps réel.
