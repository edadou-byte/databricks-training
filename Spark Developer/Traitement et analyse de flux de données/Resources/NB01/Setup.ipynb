```python
print("début copie des fichiers")
data_orders_folder = full_path_value + "data/orders"

json_files_orders = [f for f in os.listdir(data_orders_folder) if f.endswith('.json')]

print("test")

for json_file in json_files_orders:
    print(f"copie du fichier {json_file} vers le volume")
    src_path = os.path.join(data_orders_folder, json_file)
    dest_path = f"/Volumes/{catalog}/{schema}/{volume}/spark_streaming/orders/{json_file}"
    with open(src_path, "r") as f:
        dbutils.fs.put(dest_path, f.read(), overwrite=True)
```
