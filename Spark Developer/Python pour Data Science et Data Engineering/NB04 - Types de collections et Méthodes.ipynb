## Les méthodes

### Qu'est-ce qu'un objet en Python ?

En Python, **un objet** est une entité qui regroupe des données (appelées attributs) et des comportements (appelés méthodes). Tout en Python est un objet : nombres, chaînes de caractères, listes, fonctions, etc. Les objets sont créés à partir de **classes**, qui servent de "plans" pour définir la structure et le comportement des objets.

Exemple :
```python
class Chien:
    def __init__(self, nom):
        self.nom = nom

    def aboyer(self):
        print(f"{self.nom} aboie !")

mon_chien = Chien("Rex")  # mon_chien est un objet de la classe Chien
mon_chien.aboyer()        # Affiche : Rex aboie !
```

Ici, `mon_chien` est un objet qui possède un attribut (`nom`) et une méthode (`aboyer`).

### Qu'est-ce qu'une méthode en Python ?

Une **méthode** est une fonction définie à l'intérieur d'une classe et qui agit sur les objets créés à partir de cette classe. Les méthodes permettent de manipuler ou d'accéder aux données (attributs) de l'objet. Elles sont appelées en utilisant la syntaxe `objet.methode()`.

Exemple :
```python
class Voiture:
    def __init__(self, marque):
        self.marque = marque

    def klaxonner(self):
        print(f"La {self.marque} klaxonne !")

ma_voiture = Voiture("Renault")
ma_voiture.klaxonner()  # Affiche : La Renault klaxonne !
```


Ici, `klaxonner` est une méthode de la classe `Voiture`. Elle peut utiliser les attributs de l'objet (comme `self.marque`) et être appelée sur chaque instance de la classe.

### Les méthodes pour les chaînes de caractères (string) en Python

Une **méthode de string** est une fonction qui permet de manipuler ou d'interroger une chaîne de caractères. On les utilise avec la syntaxe `ma_chaine.methode()`. Exemples courants :

- `upper()`: met la chaîne en majuscules (`"abc".upper()` → `"ABC"`)
- `lower()`: met la chaîne en minuscules
- `replace(a, b)`: remplace `a` par `b` dans la chaîne
- `split(sep)`: découpe la chaîne selon le séparateur `sep`
- `strip()`: enlève les espaces au début et à la fin
- `find(sous_chaine)`: retourne l'index de la sous-chaîne (ou -1 si absent)

Ces méthodes facilitent le traitement et la transformation des textes en Python.

````python
# exemple
greeting = "Hello"
print(greeting.upper())
print(greeting)
````

### La méthode `help()` en Python

La méthode `help()` permet d'obtenir de l'aide sur un objet, une fonction, une classe ou un module en Python. Elle affiche la documentation associée, ce qui est utile pour comprendre comment utiliser une méthode ou connaître ses paramètres.

Exemple :
```python
help(str.upper)  # Affiche l'aide sur la méthode upper des chaînes de caractères
```


On peut aussi utiliser `help()` sans argument pour accéder à l'aide interactive de Python.

````python
help(greeting.capitalize)
````

````python
greeting.capitalize()
````

## Les types de colléctions

### Le type de collection `list` en Python

Une **liste** (`list`) est un type de collection en Python qui permet de stocker une séquence ordonnée d'éléments, qui peuvent être de types différents (nombres, chaînes, objets, etc.). Les listes sont modifiables (mutables), ce qui signifie qu'on peut changer, ajouter ou supprimer des éléments après leur création.

Exemple de création d'une liste :
```python
ma_liste = [1, "bonjour", 3.14, True]
```


Principales caractéristiques :
- Les éléments sont accessibles par leur index (le premier élément a l'index 0).
- On peut ajouter un élément avec `append()`, en supprimer avec `remove()` ou `pop()`, et modifier un élément en utilisant son index.
- Les listes peuvent contenir d'autres listes (listes imbriquées).

Exemple d'utilisation :
```python
fruits = ["pomme", "banane", "cerise"]
fruits.append("orange")      # Ajoute "orange" à la fin
print(fruits[1])             # Affiche "banane"
fruits[0] = "kiwi"           # Remplace "pomme" par "kiwi"
```

````python
breakfast_list = ["eggs", "bacon", "pancakes"]
print(breakfast_list)
````

````python
type(breakfast_list)
````

on peut donc ajouter un élément à une liste avec la méthode `list.append()`

````python
breakfast_list.append("coffee")
print(breakfast_list)
````

En retirer avec `list.pop()` ou `list.remove()`

````python
breakfast_list.remove("pancakes")
print(breakfast_list)
````

On peut concatener 2 liste avec l'opérateur `+`

````python
["hello", "world"] + ["from", "python"]

breakfast_list = breakfast_list + ["pancakes"]
print(breakfast_list)
````

### Les index dans les listes

Un **index** est un numéro qui permet d'accéder à un élément précis d'une liste. En Python, l'indexation commence à 0 : le premier élément a l'index 0, le deuxième l'index 1, etc.

Exemple :
```python
fruits = ["pomme", "banane", "cerise"]
print(fruits[0])  # Affiche "pomme"
print(fruits[2])  # Affiche "cerise"
```


On peut aussi utiliser des index négatifs pour accéder aux éléments en partant de la fin :
- `-1` : dernier élément
- `-2` : avant-dernier, etc.

Exemple :
python
print(fruits[-1])  # Affiche "cerise"

````python
fruits = ["pomme", "banane", "cerise"]
print(fruits[0])  # Affiche "pomme"
print(fruits[2])  # Affiche "cerise"
````

````python
print(fruits[-1])  # Affiche "cerise"
print(fruits[-2])
````

On peut également utiliser une *range* d'index : `fruits[1:2]`

````python
fruits[1:3]
````

- Si on ne définie pas d'index de départ, Python éstime que le départ est le début de la liste
- Si on ne définie pas d'index de départ, Python éstime que la fin est la fin de la liste

````python
print(fruits[:2])
print(fruits[1:])
````

On peut également modifier un élément en utilisant son index

````python
fruits[1] = 'Kiwi'
print(fruits)
````

On peut utiliser `in` pour vérifier si un élément existe

````python
print('Kiwi' in fruits)
print('Waffles' in fruits)
````

### Le type de collection `dict` en Python

Un **dictionnaire** (`dict`) est une collection en Python qui stocke des paires clé-valeur. Chaque clé est unique et permet d'accéder rapidement à sa valeur associée. Les dictionnaires sont mutables : on peut ajouter, modifier ou supprimer des éléments après leur création.

Exemple de création d'un dictionnaire :
python
mon_dict = {"nom": "Alice", "âge": 25, "ville": "Paris"}


Principales caractéristiques :
- Accès aux valeurs par leur clé : `mon_dict["nom"]` renvoie `"Alice"`.
- Ajout ou modification d'une paire : `mon_dict["pays"] = "France"`.
- Suppression d'une paire : `del mon_dict["ville"]`.
- Les clés doivent être immuables (chaînes, nombres, tuples...).

Exemple d'utilisation :
python
personne = {"nom": "Bob", "âge": 30}
print(personne["âge"])         # Affiche 30
personne["âge"] = 31           # Modifie l'âge
personne["ville"] = "Lyon"     # Ajoute une nouvelle clé

### Le type de collection `dict` en Python

Un **dictionnaire** (`dict`) est un type de collection en Python qui permet de stocker des paires **clé-valeur**. Contrairement aux listes qui utilisent des index numériques, les dictionnaires utilisent des clés (qui peuvent être des chaînes, des nombres, etc.) pour accéder aux valeurs.

Exemple de création d'un dictionnaire :
```python
mon_dict = {"nom": "Alice", "age": 30, "ville": "Paris"}
```

Principales caractéristiques :
- Les clés doivent être uniques et immuables (chaînes, nombres, tuples)
- Les valeurs peuvent être de n'importe quel type
- Les dictionnaires sont modifiables (mutables)
- L'ordre des éléments est préservé (depuis Python 3.7+)

Exemple d'utilisation :
```python
personne = {"nom": "Bob", "age": 25}
print(personne["nom"])        # Affiche "Bob"
personne["age"] = 26          # Modifie la valeur
personne["email"] = "bob@example.com"  # Ajoute une nouvelle paire clé-valeur
```

````python
# Création d'un dictionnaire
person = {"name": "Alice", "age": 30, "city": "Paris"}
print(person)
print(type(person))
````

````python
# Accéder aux valeurs avec les clés
print(person["name"])
print(person["age"])
````

````python
# Modifier une valeur existante
person["age"] = 31
print(person)

# Ajouter une nouvelle paire clé-valeur
person["email"] = "alice@example.com"
print(person)
````

### Méthodes utiles des dictionnaires

Quelques méthodes courantes :
- `dict.keys()` : retourne toutes les clés
- `dict.values()` : retourne toutes les valeurs
- `dict.items()` : retourne les paires clé-valeur
- `dict.get(clé)` : retourne la valeur associée à la clé (sans erreur si la clé n'existe pas)
- `dict.pop(clé)` : supprime et retourne la valeur associée à la clé
- `dict.update(autre_dict)` : fusionne deux dictionnaires

````python
print(person.get("name"))
print(person.get("email"))
print(person.get("city"))
````
