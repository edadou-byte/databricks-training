# Librairies

En Python, une **librairie** (ou bibliothèque) est un ensemble de modules contenant des fonctions, des classes et des outils prêts à l’emploi pour faciliter le développement. Les librairies permettent de réutiliser du code existant pour accomplir des tâches courantes comme la manipulation de données, l’analyse mathématique, la création de graphiques, ou l’accès à des bases de données.

Par exemple :
- `numpy` pour le calcul scientifique,
- `pandas` pour la manipulation de données,
- `matplotlib` pour la visualisation,
- `requests` pour les requêtes HTTP.

Pour utiliser une librairie, il faut généralement l’installer (avec `pip install nom_de_la_librairie`) puis l’importer dans le code avec `import nom_de_la_librairie`.

## La commande `%run` dans Databricks

La commande **`%run`** est une commande magique spécifique à Databricks qui permet d'exécuter un autre notebook et d'importer toutes ses variables, fonctions et classes dans le notebook actuel.

### Syntaxe
```python
%run /chemin/vers/le/notebook
```

### Caractéristiques principales

* **Réutilisation de code** : Permet de centraliser des fonctions communes, des configurations ou des définitions de variables dans un notebook séparé
* **Partage de contexte** : Toutes les variables, fonctions et imports du notebook exécuté deviennent disponibles dans le notebook appelant
* **Chemins** : Peut utiliser des chemins absolus (`/Users/email/mon_notebook`) ou relatifs (`./mon_notebook`)
* **Pas d'extension** : Ne pas inclure l'extension `.py` ou `.ipynb` dans le chemin

### Exemple d'utilisation

**Notebook de configuration** (`/Shared/config`):
```python
# Variables de configuration
DATA_PATH = "/mnt/data/"
MAX_RECORDS = 1000

def nettoyer_donnees(df):
    return df.dropna()
```

**Notebook principal**:
```python
%run /Shared/config

# Utilisation des variables et fonctions importées
df = spark.read.parquet(DATA_PATH)
df_clean = nettoyer_donnees(df)
```

### Différence avec `import`

* **`%run`** : Exécute tout le code du notebook et partage le contexte complet
* **`import`** : Importe uniquement des modules Python spécifiques sans exécuter tout le code

## PyPi et les bibliothèques

**PyPi** (Python Package Index) est le principal dépôt en ligne de packages Python. Il permet aux développeurs de publier, partager et installer facilement des bibliothèques et outils Python via la commande `pip`. Une **bibliothèque** (ou librairie) est un ensemble de modules réutilisables qui facilitent le développement en fournissant des fonctions, classes ou outils prêts à l’emploi pour des tâches variées (analyse de données, web, machine learning, etc.). Pour utiliser une bibliothèque, on l’installe généralement depuis PyPi avec `pip install nom_de_la_bibliothèque`, puis on l’importe dans son code avec `import`.

### pip

**pip** est l’outil officiel de gestion de paquets pour Python. Il permet d’installer, de mettre à jour et de désinstaller des bibliothèques Python depuis le dépôt en ligne PyPi (Python Package Index). Pour installer une bibliothèque, il suffit d’utiliser la commande `pip install nom_de_la_bibliothèque` dans le terminal. pip facilite ainsi l’ajout de fonctionnalités à vos projets Python en quelques commandes simples.

   
### La fonction `help()`

La fonction intégrée `help()` en Python permet d’obtenir de l’aide sur une fonction, une classe, un module ou un objet. Elle affiche la documentation (docstring) associée à l’élément demandé.

**Exemples d’utilisation :**
```python
help(print)           # Affiche l’aide sur la fonction print
help(str)             # Affiche l’aide sur la classe str
help(numpy.array)     # Affiche l’aide sur la fonction array du module numpy
```


Dans un notebook, on peut aussi utiliser le point d’interrogation :
```python
print?                # Affiche l’aide sur print
```

````python
help(np.sqrt)
````

````python
sqrt?
````
