### Filtrage (filtres) sur un DataFrame pandas

Le filtrage dans pandas permet de sélectionner des lignes ou des colonnes selon des critères spécifiques. Voici les principales méthodes :

- **Filtrer les lignes selon une condition**  
  On utilise des expressions booléennes :
  ```python
  # Sélectionner les lignes où la colonne 'age' est supérieure à 30
  df_filtre = df[df['age'] > 30]
  ```
  

- **Filtrer plusieurs conditions**  
  On combine les conditions avec `&` (et) ou `|` (ou) :
  ```python
  # Lignes où 'age' > 30 ET 'sexe' == 'F'
  df_filtre = df[(df['age'] > 30) & (df['sexe'] == 'F')]
  ```
  

- **Filtrer des colonnes par nom**  
  On sélectionne une ou plusieurs colonnes :
  ```python
  # Sélectionner uniquement les colonnes 'nom' et 'age'
  df_filtre = df[['nom', 'age']]
  ```
  

- **Méthode `filter()`**  
  Pour filtrer selon les noms de colonnes ou d'index :
  ```python
  # Colonnes dont le nom contient 'date'
  df_filtre = df.filter(like='date', axis=1)
  ```
  

- **Méthode `query()`**  
  Pour filtrer avec une syntaxe proche du SQL :
  ```python
  # Lignes où 'age' > 30
  df_filtre = df.query('age > 30')
  ```
  

Le filtrage est essentiel pour explorer, nettoyer et analyser efficacement vos données avec pandas.

## Les opérateurs booléens dans pandas

Les opérateurs booléens permettent de filtrer et de manipuler les données selon des conditions logiques. Voici les principaux opérateurs utilisés dans pandas :

- **`==`** : égal à  
  `df[df['colonne'] == valeur]`

- **`!=`** : différent de  
  `df[df['colonne'] != valeur]`

- **`>`** : supérieur à  
  `df[df['colonne'] > valeur]`

- **`<`** : inférieur à  
  `df[df['colonne'] < valeur]`

- **`>=`** : supérieur ou égal à  
  `df[df['colonne'] >= valeur]`

- **`<=`** : inférieur ou égal à  
  `df[df['colonne'] <= valeur]`

Pour combiner plusieurs conditions :
- **`&`** : ET logique (and)  
  `df[(df['col1'] > 10) & (df['col2'] == 'A')]`

- **`|`** : OU logique (or)  
  `df[(df['col1'] > 10) | (df['col2'] == 'A')]`

- **`~`** : NON logique (not)  
  `df[~(df['col1'] > 10)]`

**Attention :**  
Les conditions doivent être entourées de parenthèses lors de la combinaison avec `&`, `|` ou `~`.

Ces opérateurs sont essentiels pour filtrer, sélectionner ou exclure des lignes selon des critères précis dans un DataFrame pandas.

### Fonctions d'agrégation en pandas

Les fonctions d'agrégation permettent de résumer ou de calculer des statistiques sur un ensemble de données. Elles sont souvent utilisées après un regroupement (`groupby`) pour obtenir des informations synthétiques.

Voici les principales fonctions d'agrégation en pandas :

- **sum()** : somme des valeurs
- **mean()** : moyenne
- **min() / max()** : minimum / maximum
- **count()** : nombre d'éléments non nuls
- **median()** : médiane
- **std()** : écart-type
- **agg()** : permet d'appliquer plusieurs fonctions d'agrégation à la fois

**Exemple d'utilisation :**

```python
import pandas as pd

df = pd.DataFrame({
    'groupe': ['A', 'A', 'B', 'B'],
    'valeur': [10, 20, 30, 40]
})

# Moyenne par groupe
df.groupby('groupe')['valeur'].mean()

# Plusieurs agrégations
df.groupby('groupe')['valeur'].agg(['sum', 'mean', 'max'])
```


Ces fonctions sont essentielles pour analyser et résumer rapidement vos données.

Dans un prémier temps on va rajouter 2 colonnes au Dataset :

- price

````python
print(df["review_number"].mean())
print(df["review_number"].min())
print(df["review_number"].max())
````

### Le groupby avec pandas

La méthode `groupby()` de pandas permet de regrouper les données d'un DataFrame selon une ou plusieurs colonnes, puis d'appliquer des fonctions d'agrégation (comme la somme, la moyenne, le comptage, etc.) sur chaque groupe.

**Principe :**
1. On "split" les données selon une ou plusieurs colonnes (ex : par marque, par région, etc.).
2. On "apply" une fonction d'agrégation sur chaque groupe.
3. On "combine" les résultats pour obtenir un DataFrame résumé.

**Exemple :**

```python
import pandas as pd

df = pd.DataFrame({
    'marque': ['Apple', 'Samsung', 'Apple', 'Honor', 'Samsung'],
    'prix': [1000, 800, 1200, 600, 900]
})

# Moyenne du prix par marque
df.groupby('marque')['prix'].mean()
```



On peut aussi appliquer plusieurs fonctions d'agrégation avec `agg()` :

```python
df.groupby('marque')['prix'].agg(['mean', 'max', 'min'])
```

Le groupby est très utile pour analyser et résumer rapidement des données par catégorie.

### Réinitialiser l'index avec pandas : `reset_index()`

La méthode `reset_index()` permet de remettre à zéro l'index d'un DataFrame ou d'une Series, c'est-à-dire de remplacer l'index actuel par un index numérique par défaut (0, 1, 2, ...). Cette opération est utile après des opérations de filtrage, de groupby ou lorsque l'index n'a plus de signification.

- Par défaut, l'ancien index devient une colonne du DataFrame.
- Avec `drop=True`, l'ancien index est supprimé et non ajouté comme colonne.
- Avec `inplace=True`, la modification se fait directement sur le DataFrame sans en créer un nouveau.

**Exemples :**
```python
import pandas as pd

df = pd.DataFrame({'A': [10, 20, 30]}, index=['x', 'y', 'z'])

# Réinitialiser l'index (l'ancien index devient une colonne)
df_reset = df.reset_index()
print(df_reset)
#    index   A
# 0      x  10
# 1      y  20
# 2      z  30

# Réinitialiser l'index sans conserver l'ancien index
df_reset2 = df.reset_index(drop=True)
print(df_reset2)
#     A
# 0  10
# 1  20
# 2  30
```

### Trier un DataFrame avec pandas

Le tri (ou "sort") dans pandas permet de réorganiser les lignes d'un DataFrame selon la valeur d'une ou plusieurs colonnes. On utilise la méthode `sort_values()`.

**Exemple de base :**
```python
import pandas as pd

df = pd.DataFrame({
    'nom': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 20]
})

# Trier par la colonne 'age' (ordre croissant)
df_sorted = df.sort_values(by='age')
print(df_sorted)
```


**Résultat :**

|   | nom     | age |
|---|---------|-----|
| 2 | Charlie |  20 |
| 0 | Alice   |  25 |
| 1 | Bob     |  30 |


- Pour trier en ordre décroissant, ajoutez `ascending=False` :
  ```python
  df.sort_values(by='age', ascending=False)
  ```

- Pour trier selon plusieurs colonnes :
  ```python
  df.sort_values(by=['age', 'nom'])
  ```

Le tri est utile pour organiser, visualiser ou préparer les données avant analyse.

````python
display(df.sort_values(["price"]).head(3))
````

Par défaut le tri est ascendant, et si on l'applique à une `Series` il n'y à pas besoin de spécifier une colonne :

````python
df["price"].sort_values()
````

Pour trier en descendant on utilise le paramètre `ascending=False` : 

````python
df["price"].sort_values(ascending=False)
````

## Le NaN dans un DataFrame pandas

Dans un DataFrame pandas, la valeur `NaN` (Not a Number) représente une donnée manquante ou absente. Elle est utilisée pour indiquer qu'une cellule ne contient pas de valeur valide, par exemple lorsqu'une information n'est pas disponible dans le jeu de données.

- `NaN` est souvent généré lors de la lecture de fichiers incomplets ou après des opérations de jointure ou de transformation.
- pandas fournit des méthodes pour détecter (`isna()`, `isnull()`), filtrer et traiter les valeurs `NaN` (par exemple avec `fillna()` pour remplacer ou `dropna()` pour supprimer).

Ici, les `NaN` indiquent des valeurs manquantes dans les colonnes A et B.

On rajoute une colonne **paid** (aléatoire, qui représente si le téléphone à été payé en une fois) au Dataframe. 

````python
df["paid"] = np.random.choice([True, False, np.nan], size=len(df))
display(df.head())
````

````python
nan_df = df[["paid"]]
display(nan_df.head())
````

````python
nan_df.isna().sum()
````

Supprimons toutes les lignes avec `NaN` comme valeur dans la colonne **paid** :

````python
nan_df_temp = df

nan_df_temp = nan_df_temp.dropna()
display(nan_df_temp.head())
nan_df_temp.isna().sum()
print(f"number of lines after dropping NaN values : {nan_df_temp.count()}")
````

On peut également remplacer toutes les valeurs `NaN`, pour éviter des pettre des données (car avec le drop les lignes entières sont retirées du Dataframe) dans ce cas remplacons par 0 (False) : 

````python
nan_df = df

print(f"number of NaN values : {nan_df.isna().sum()}")
print(f"Number of False values : {nan_df[nan_df["paid"] == False].count()}")
print(f"Number of True values : {nan_df[nan_df["paid"] == True].count()}")

nan_df.fillna(0, inplace=True)

print(f"Number of False values after fillna() : {nan_df[nan_df["paid"] == False].count()}")
````
