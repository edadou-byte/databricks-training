### Les DataFrames Pandas

Un DataFrame est la structure de données principale de la bibliothèque Pandas. Il s’agit d’un tableau à deux dimensions (lignes et colonnes), similaire à une feuille de calcul Excel ou une table SQL. Chaque colonne peut contenir un type de données différent (nombres, chaînes de caractères, dates, etc.).

Les principales caractéristiques des DataFrames Pandas sont :
- **Indexation** : chaque ligne et colonne possède un index, facilitant l’accès et la manipulation des données.
- **Flexibilité** : possibilité d’ajouter, supprimer ou modifier des colonnes et des lignes facilement.
- **Opérations puissantes** : filtrage, tri, agrégation, fusion, gestion des valeurs manquantes, etc.
- **Compatibilité** : lecture et écriture dans de nombreux formats (CSV, Excel, SQL, JSON…).

Exemple de création d’un DataFrame :
```python
import pandas as pd

data = {'Nom': ['Alice', 'Bob', 'Charlie'],
        'Âge': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)
```

````python
data = [["Emile", 29, "Developer"], ["Johann", 38, "Tech Lead"], ["Davy", 43, "Manager"]]

df = pd.DataFrame(data)
df
````

Pour ajouter des noms de colonnes à un DataFrame Pandas, utilisez le paramètre `columns` lors de la création du DataFrame :

```python
data = [["Emile", 29, "Developer"], ["Johann", 38, "Tech Lead"], ["Davy", 43, "Manager"]]
colonnes = ["Nom", "Âge", "Poste"]

df = pd.DataFrame(data, columns=colonnes)
print(df)
```

````python
data = [["Emile", 29, "Developer"], ["Johann", 38, "Tech Lead"], ["Davy", 43, "Manager"]]
colonnes = ["Nom", "Age", "Poste"]

df = pd.DataFrame(data, columns=colonnes)
df
````

### Les Series Pandas

Une **Series** est une structure de données unidimensionnelle de Pandas, similaire à une colonne dans un tableau ou une feuille Excel. Chaque valeur d’une Series est associée à un index, ce qui permet d’accéder facilement aux éléments.

Caractéristiques principales :
- Peut contenir tout type de données (nombres, chaînes, booléens, etc.)
- L’index peut être personnalisé ou généré automatiquement
- Prend en charge les opérations vectorisées et le filtrage

Exemple de création d’une Series :
```python
import pandas as pd

ages = pd.Series([25, 30, 35], index=["Alice", "Bob", "Charlie"])
print(ages)
```

On peut recupérer une **Series** de deux manières :
- `df["nom_colonne"]`
- `df.nom_colonne`

````python
df["Age"]
````

````python
df.Age
````

### Les dtypes dans Pandas

Le terme **dtype** (data type) désigne le type de données de chaque colonne dans un DataFrame ou d’une Series Pandas. Les dtypes permettent à Pandas de savoir comment stocker et manipuler efficacement les données.

Quelques exemples de dtypes courants :
- `int64` : entiers
- `float64` : nombres décimaux
- `object` : chaînes de caractères ou objets Python
- `bool` : booléens (True/False)
- `datetime64[ns]` : dates et heures

Connaître les dtypes est important pour :
- Optimiser la mémoire utilisée
- Éviter des erreurs lors de calculs ou de manipulations
- S’assurer que les opérations sont adaptées au type de données

Pour afficher les dtypes d’un DataFrame :
```python
print(df.dtypes)
```
Cela retourne le type de chaque colonne du DataFrame.

Si on regarde une **Series** on peut voir en dessous dtype : int64

````python
df["Age"]
````

On peut vérifier tout les dtypes des colonnes d'un Dataframe avec l'attribut `dtypes`

````python
df.dtypes
````

### Opérations sur les Series Pandas

Les **Series** Pandas permettent d’effectuer facilement des opérations vectorisées, c’est-à-dire des calculs appliqués à l’ensemble des éléments sans boucle explicite. Voici quelques exemples d’opérations courantes :

- **Opérations arithmétiques** : addition, soustraction, multiplication, division, etc. Exemple : `ages + 1` ajoute 1 à chaque valeur de la Series.
- **Opérations de comparaison** : égalité, supériorité, etc. Exemple : `ages > 30` retourne une Series de booléens.
- **Méthodes statistiques** : `mean()`, `sum()`, `min()`, `max()`, etc.
- **Méthodes de transformation** : `apply()`, `map()`, `astype()`, etc.
- **Gestion des valeurs manquantes** : `isnull()`, `fillna()`, etc.
- **Accès aux éléments** : par index ou par position (`ages["Alice"]`, `ages.iloc[0]`).

Ces opérations sont rapides et efficaces grâce à l’optimisation interne de Pandas.

Par exemple additionner 2 `int64 Series` revient à additionner deux int en python

````python
df["Age"] + df["Age"]
````

````python
df["Age"] * 3
````

#### Sélectionner des valeurs dans une Series

Pour sélectionner des valeurs dans une **Series** Pandas, on peut utiliser plusieurs méthodes :

- **Par label et index** : 
  Utilisez `df["nom_colonne"][index]`
- **Par position** :  
  Utilisez `.iloc[]` pour accéder à une valeur par sa position (indice entier).  
  Exemple :  
  ```python
  ages.iloc[0]  # Première valeur de la Series
  ```
  

- **Par label** :  
  Utilisez `.loc[]` pour accéder à une valeur par son label d’index.  
  Exemple :  
  ```python
  ages.loc["Alice"]  # Valeur associée au label "Alice"
  ```
  

- **Par masque booléen** :  
  Sélectionnez plusieurs valeurs selon une condition.  
  Exemple :  
  ```python
  ages[ages > 30]  # Toutes les valeurs supérieures à 30
  ```
  

- **Avec la méthode `.get()`** :  
  Permet de récupérer une valeur par label, avec une valeur par défaut si le label n’existe pas.  
  Exemple :  
  ```python
  ages.get("Bob", default=-1)
  ```
  

- **Avec la méthode `.take()`** :  
  Permet de récupérer des valeurs par positions multiples.  
  Exemple :  
  ```python
  ages.take([0, 2, 4])
  ```
