# Types de cellules

Un notebook Databricks est une interface web interactive permettant de créer, exécuter et documenter des workflows de data science et de machine learning. Il prend en charge plusieurs langages (Python, SQL, Scala, R) dans un même document, facilite la collaboration en temps réel, la visualisation des données et le versioning automatique. Les notebooks peuvent contenir du texte formaté (Markdown), du code exécutable, des visualisations et des résultats, offrant ainsi un environnement complet pour l'analyse et le développement de projets de données.

- Markdown
- Code (Python, SQL, Scala R)

# Comment utiliser le markdown dans Databricks

Le markdown permet de formater du text, ajouter des images, faires des listes, ... dans un notebook databricks

## Syntaxe du markdown

- **gras:** `**text**` → **text**
- *Italique:* `*text*` → *text*
- [Lien](https://databricks.com): `[Link](https://databricks.com)`
- Lists:
  - à point: `- item`
  - à numéro: `1. item`
- Code: `` `code` `` or bloc de code:
  
  print("Hello, Markdown!")
  
- Titres: `#`, `##`, `###` pour différents niveau

> Rentre `%md` au début de la cellule pour utiliser la markdown

## Cellules Python dans Databricks

Les cellules Python permettent d'exécuter du code Python directement dans le notebook Databricks. Pour créer une cellule Python, il suffit d'écrire du code Python ou de commencer la cellule par `%python`. Ces cellules sont utilisées pour manipuler des données, entraîner des modèles, visualiser des résultats, etc.

Exemple :
```python
# Calcul simple
x = 5
y = 10
print(x + y)
```


Les résultats s'affichent directement sous la cellule après exécution.

````python
print('hello from databricks')
# Calcul simple
x = 5
y = 10
print(x + y)
````

## Cellules SQL dans Databricks

Les cellules SQL permettent d'exécuter des requêtes SQL directement dans le notebook Databricks pour interroger, manipuler et analyser des données stockées dans des tables. Pour créer une cellule SQL, commencez la cellule par `%sql` ou sélectionnez le langage SQL dans l'éditeur.

Exemple :
```sql
%sql
SELECT * FROM ma_table LIMIT 10
```


Les résultats des requêtes s'affichent sous la cellule, avec des options de visualisation (tableau, graphique, etc.). Les cellules SQL sont particulièrement utiles pour l'exploration rapide des données et la création de rapports.

````python
%sql
SELECT current_date;
````
