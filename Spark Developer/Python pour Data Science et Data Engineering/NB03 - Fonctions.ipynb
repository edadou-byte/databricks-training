### Résumé des fonctions en Python

Une fonction en Python est un bloc de code réutilisable qui effectue une tâche spécifique. On la définit avec le mot-clé `def`, suivi du nom de la fonction et de parenthèses pouvant contenir des paramètres.

**Syntaxe de base :**
python
def nom_de_la_fonction(param1, param2):
    # instructions
    return resultat


- **Appel** : On exécute une fonction en écrivant son nom suivi de parenthèses : `nom_de_la_fonction(arg1, arg2)`
- **Paramètres** : Permettent de passer des valeurs à la fonction.
- **Valeur de retour** : Le mot-clé `return` permet de renvoyer un résultat.

**Exemple :**
```python
def addition(a, b):
    return a + b

resultat = addition(3, 5)  # resultat vaut 8
```

Quand python exécute du code et voit un appel à une fonction, il saute à l'intérieur du bloc de code de la fonction et une fois que cela à finis sort et reprend au niveau de l'appel.

````python
# Exemple
def ten_dollars_to_euros():
    print(10.0 * 0.93)

ten_dollars_to_euros()
````

#### Les paramètres dans une fonction Python

Les **paramètres** d'une fonction sont des variables placées entre les parenthèses lors de la définition de la fonction. Ils servent à transmettre des valeurs à la fonction lors de son appel.

- **Paramètre** : variable définie dans la déclaration de la fonction.
- **Argument** : valeur réelle passée à la fonction lors de l'appel.

Exemple :
```python
def saluer(nom):
    print("Bonjour", nom)

saluer("Alice")  # "Alice" est l'argument passé au paramètre nom
```


Il existe plusieurs types de paramètres :
- **Obligatoires** : doivent être fournis lors de l'appel.
- **Optionnels (avec valeur par défaut)** : prennent une valeur si aucun argument n'est fourni.
- **Paramètres multiples** : `*args` pour un nombre variable d'arguments positionnels, `**kwargs` pour des arguments nommés.

Exemple :
```python
def exemple(a, b=2, *args, **kwargs):
    pass
```

**Exemple** : fonction qui transforme des dollars en euros

Dans ce cas :
- *dollar_amount* est le paramètre de la fonction
- *9, 10, 20.0* : sont les arguments (valeurs réelles) passés à cette fonction

````python
def dollars_to_euros(dollar_amount):
    print(dollar_amount * 0.93)

dollars_to_euros(9)
dollars_to_euros(10)
dollars_to_euros(20.0)
````

On peut également utiliser plusieurs paramètres (et arguments) dans une fonctions, il faut les séparer dans la déclaration de la fonction : 
```python
def test(val1, val2)
  print(val1 * val2)

test(10, 2) # affiche 20
```

Continuons l'exemple précédents en passant la taux de conversion en paramètre de la fonction

````python
def dollars_to_euros_with_rate(dollar_amount, rate):
    print(dollar_amount * rate)

dollars_to_euros_with_rate(9, 0.93)
dollars_to_euros_with_rate(10, 1)
dollars_to_euros_with_rate(20.0) # cela causera une erreur
````

#### Invocation nommée en Python

L'**invocation nommée** (ou appel avec arguments nommés) consiste à préciser le nom des paramètres lors de l'appel d'une fonction. Cela permet de passer les arguments dans n'importe quel ordre et rend le code plus lisible.

**Exemple :**
```python
def saluer(prenom, message):
    print(f"{message}, {prenom}!")

saluer(message="Bonjour", prenom="Alice")  # affiche "Bonjour, Alice!"
```

Ici, `message="Bonjour"` et `prenom="Alice"` sont des arguments nommés.  
Cela évite les erreurs d'ordre et rend l'appel explicite.

````python
# appliquons sur la fonction précédente
dollars_to_euros_with_rate(dollar_amount=20.0, rate=0.93)
dollars_to_euros_with_rate(rate=0.93, dollar_amount=20.0)
````

#### Valeur par défaut d'un paramètre

Une **valeur par défaut** permet à un paramètre de prendre automatiquement une valeur si aucun argument n'est fourni lors de l'appel de la fonction. On l'indique en assignant une valeur au paramètre dans la déclaration de la fonction.

**Exemple :**
```python
def saluer(nom, message="Bonjour"):
    print(f"{message}, {nom}!")

saluer("Alice")           # affiche "Bonjour, Alice!"
saluer("Bob", "Salut")    # affiche "Salut, Bob!"
```

Ici, `message` a pour valeur par défaut `"Bonjour"`. Si on ne précise pas `message` lors de l'appel, cette valeur sera utilisée.

````python
def saluer(nom, message="Bonjour"):
    print(f"{message}, {nom}!")

saluer("Alice")           # affiche "Bonjour, Alice!"
saluer("Bob", "Salut")    # affiche "Salut, Bob!"
````

#### Output / return d'une fonction

Le mot-clé `return` permet à une fonction de renvoyer une valeur à l'endroit où elle a été appelée. Lorsqu'une fonction rencontre `return`, elle s'arrête immédiatement et transmet la valeur spécifiée.

**Exemple :**
```python
def carre(x):
    return x * x

resultat = carre(4)  # resultat vaut 16
```


Si aucune instruction `return` n'est présente, la fonction renvoie automatiquement `None`.
```python
def afficher():
    print("Bonjour")

val = afficher()  # val vaut None
```

````python
print(dollars_to_euros_with_rate(90, 1))
````

reprenons l'exemple sur la conversion du dollars et plutôt que de faire un `print()` dans la fonction, la fonction retournera la valeur calculée

````python
def dollars_to_euros_with_rate(dollar_amount, rate):
    return dollar_amount * rate

euros = dollars_to_euros_with_rate(9, 0.93)
print(euros)
````

### Les type hints en Python

Les **type hints** (annotations de type) permettent d’indiquer le type attendu des paramètres et de la valeur de retour d’une fonction. Ils servent à documenter le code, faciliter la relecture, et améliorer l’autocomplétion dans les éditeurs.

**Syntaxe :**
```python
def addition(a: int, b: int) -> int:
    return a + b
```
Ici, `a` et `b` sont des entiers (`int`), et la fonction retourne un entier.

Les type hints ne changent pas le comportement du code à l’exécution, mais ils aident à détecter des erreurs plus tôt avec des outils comme `mypy` ou les IDE.

**Exemple avec types complexes :**
```python
from typing import List

def moyenne(notes: List[float]) -> float:
    return sum(notes) / len(notes)
```

````python
# exemple avec la fonction dollars_to_euros_with_rate
def dollars_to_euros_with_rate(dollar_amount: float, rate: float):
    return dollar_amount * rate

a = dollars_to_euros_with_rate(9, 0.93)
print(a)
````

#### Les docstrings en Python

Une **docstring** (documentation string) est une chaîne de caractères placée juste après la définition d’une fonction, d’une classe ou d’un module. Elle sert à documenter le code et à expliquer son fonctionnement.

**Syntaxe :**
```python
def addition(a, b):
    """
    Additionne deux nombres et retourne le résultat.
    
    Args:
        a (int): premier nombre
        b (int): second nombre

    Returns:
        int: la somme de a et b
    """
    return a + b
```


La docstring est accessible via l’attribut `__doc__` de la fonction :
python
print(addition.__doc__)


Les docstrings facilitent la compréhension du code et sont utilisées par les outils d’aide intégrée (`help()`, IDE, documentation automatique).

````python
# exemple avec la fonction dollars_to_euros_with_rate
def dollars_to_euros_with_rate(dollar_amount: float, rate: float):
    """
    Retourne le montant en euros correspondant à un montant en dollars et à un taux de change
    Paramètres :
        dollar_amount (float) : valeur à convertir
        rate (float) : taux de change
    Returns
        euro_amount (float) : valeur convertie
    """
    euro_amount = dollar_amount * rate
    return euro_amount

a = dollars_to_euros_with_rate(9, 0.93)
print(a)
````

````python
help(dollars_to_euros_with_rate)
````

#### Portée (scope) des variables en Python

La **portée** (scope) d’une variable détermine où cette variable est accessible dans le code.

- **Variable globale** : définie en dehors de toute fonction, accessible partout dans le module.
- **Variable locale** : définie à l’intérieur d’une fonction, accessible uniquement dans cette fonction.

**Exemple :**
```python
x = 10  # variable globale

def ma_fonction():
    y = 5  # variable locale
    print(x)  # x est accessible ici
    print(y)  # y est accessible ici

ma_fonction()
print(x)  # x est accessible ici
print(y)  # Erreur : y n'est pas accessible ici
```


Pour modifier une variable globale à l’intérieur d’une fonction, on utilise le mot-clé `global` :
```python
compteur = 0

def incrementer():
    global compteur
    compteur += 1
```


Il existe aussi le mot-clé `nonlocal` pour accéder à une variable d’une fonction englobante (cas des fonctions imbriquées).

**Résumé :**
- Les variables créées dans une fonction sont locales à cette fonction.
- Les variables créées en dehors des fonctions sont globales.
- On utilise `global` ou `nonlocal` pour modifier des variables extérieures à la fonction.

````python
x = 10  # variable globale

def ma_fonction():
    y = 5  # variable locale
    print(x)  # x est accessible ici
    print(y)  # y est accessible ici

ma_fonction()
print(x)  # x est accessible ici
print(y)  # Erreur : y n'est pas accessible ici
````

### Résumé des fonctions built-in en Python

Python propose de nombreuses **fonctions intégrées** (built-in) utilisables sans importation de module. Elles facilitent les opérations courantes sur les types de base.

Quelques exemples courants :

- `print()` : affiche une valeur à l’écran.
- `len(obj)` : retourne la longueur d’un objet (liste, chaîne, etc.).
- `type(obj)` : donne le type d’un objet.
- `int()`, `float()`, `str()` : convertissent en entier, flottant ou chaîne.
- `sum(iterable)` : fait la somme des éléments d’un itérable.
- `max(iterable)`, `min(iterable)` : valeur maximale/minimale d’un itérable.
- `range(start, stop, step)` : génère une séquence de nombres.
- `list()`, `dict()`, `set()`, `tuple()` : créent des collections.
- `input()` : lit une entrée utilisateur.
- `abs(x)` : valeur absolue.
- `round(x, n)` : arrondit un nombre.
