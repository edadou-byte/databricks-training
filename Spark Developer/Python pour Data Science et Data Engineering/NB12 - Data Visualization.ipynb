## Visualisation de données avec pandas

Pandas propose des fonctions intégrées pour visualiser rapidement les données, notamment via `.plot()`.

Exemple :

```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemple de DataFrame
df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [4, 3, 2, 1]
})

# Tracer un graphique en ligne
df.plot()
plt.show()
```

Vous pouvez également créer des histogrammes, des graphiques à barres, etc. avec `df.plot.hist()`, `df.plot.bar()`, etc.

````python
# on utiliser head(10) pour que ce soit plus visible
mean_price_by_brand = df.head(10).groupby('brand')['price'].mean().reset_index()

# Affichage de l'histogramme avec brand sur l'axe x
mean_price_by_brand.plot.bar(x='brand', y='price', legend=False, title='Moyenne de price par brand')
````

## Seaborn

Seaborn est une bibliothèque de visualisation de données basée sur matplotlib, conçue pour rendre la création de graphiques statistiques en Python plus simple et plus esthétique. Elle propose des fonctions de haut niveau pour générer rapidement des graphiques courants (histogrammes, boxplots, heatmaps, etc.) et permet de personnaliser facilement l'apparence des visualisations.

Exemple d'utilisation :

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Exemple de DataFrame
import pandas as pd
df = pd.DataFrame({
    'marque': ['A', 'B', 'A', 'B'],
    'prix': [10, 15, 12, 18]
})

# Création d'un graphique à barres du prix moyen par marque
sns.barplot(x='marque', y='prix', data=df)
plt.show()
```

Sur Databricks, il est aussi possible d'afficher les graphiques seaborn directement dans le notebook avec la fonction `display()` :

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
ax = sns.barplot(x='marque', y='prix', data=df)
fig = ax.get_figure()
display(fig)
```
