### Type 1 : Integers
Integers ou int sont des nombres entiers non décimaux

**Data** : valeurs (ex : -2, -1, 0, 1, 2, 3, ...)

**Opérations possibles** : +, -, *, / 

````python
# exemple
2 + 3 - 1 * 2
````

### Type 2 : Float
Float ou float sont des nombres à virgule flottante (décimaux).

**Data** : valeurs (ex : -2.5, 0.0, 3.14, 1e-3, 2.0)

**Opérations possibles** : +, -, *, /, ** (puissance), % (modulo)

````python
# exemple
1.2 * 3.8 + 9.1
````

pour vérifier le type d'une variable, on utilise la fonction `type()`

````python
type(6.2)
````

python :

## Les Variables en Python

Une **variable** est un conteneur qui permet de stocker une valeur en mémoire pour la réutiliser plus tard dans le programme.

### Caractéristiques principales :

* **Déclaration simple** : Pas besoin de déclarer le type, Python le détecte automatiquement
* **Affectation** : On utilise le signe `=` pour assigner une valeur à une variable
* **Nommage** : Le nom doit commencer par une lettre ou underscore, peut contenir des lettres, chiffres et underscores
* **Sensible à la casse** : `variable` et `Variable` sont deux variables différentes

### Exemples :

```python
# Variables numériques
age = 25                    # int
prix = 19.99               # float

# Variables textuelles
nom = "Alice"              # str
ville = 'Paris'            # str

# Variables booléennes
est_actif = True           # bool
est_majeur = (age >= 18)  # bool

# Réaffectation
x = 10
x = x + 5                  # x vaut maintenant 15
x += 3                     # raccourci, x vaut 18
```

### Bonnes pratiques :

* Utiliser des noms descriptifs : `prix_total` plutôt que `pt`
* Respecter la convention snake_case : `mon_age` plutôt que `monAge`
* Éviter les mots réservés Python : `class`, `for`, `if`, etc.

````python
a = 3
b = 2
c = a * b
c
````

## F-String Formatting

Les **f-strings** (formatted string literals) permettent d'insérer des variables et expressions directement dans une chaîne de caractères en préfixant la chaîne avec `f` ou `F`.

### Syntaxe de base :
```python
f"texte {variable} texte"
```

### Avantages :
* Plus lisible et concis que `.format()` ou `%`
* Évaluation d'expressions directement dans les accolades
* Meilleur performance

### Exemples :
```python
# Variables simples
nom = "Alice"
age = 25
print(f"Je m'appelle {nom} et j'ai {age} ans")
# Résultat : Je m'appelle Alice et j'ai 25 ans

# Expressions
prix = 19.99
quantite = 3
print(f"Total : {prix * quantite} euros")
# Résultat : Total : 59.97 euros

# Formatage numérique
pi = 3.14159
print(f"Pi arrondi : {pi:.2f}")  # 2 décimales
# Résultat : Pi arrondi : 3.14

# Alignement et largeur
print(f"{'Gauche':<10} | {'Centre':^10} | {'Droite':>10}")
# Résultat : Gauche     |   Centre   |     Droite
```

````python
# Exemples de f-strings
nom = "Alice"
age = 25
print(f"Je m'appelle {nom} et j'ai {age} ans")

# Avec expressions
prix = 19.99
quantite = 3
print(f"Total : {prix * quantite} euros")

# Formatage numérique
pi = 3.14159
print(f"Pi arrondi : {pi:.2f}")

# Avec des variables du notebook
print(f"\nVariables existantes : a={a}, b={b}, c={c}")
````
