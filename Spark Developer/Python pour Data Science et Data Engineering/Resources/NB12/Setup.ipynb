```python
import random
from datetime import datetime, timedelta
import pandas as pd

categories = ['Electronics', 'Clothing', 'Books', 'Toys', 'Groceries']
regions = ['North', 'South', 'East', 'West']
start_date = datetime(2025, 1, 1)

data = []
for i in range(100):
    row = (
        (start_date + timedelta(days=random.randint(0, 60))).strftime('%Y-%m-%d'),
        random.choice(categories),
        random.choice(regions),
        random.randint(1, 20),  # quantity
        round(random.uniform(5, 500), 2)  # price
    )
    data.append(row)

columns = ['date', 'category', 'region', 'quantity', 'price']

df_test = pd.DataFrame(data, columns=columns)
```
