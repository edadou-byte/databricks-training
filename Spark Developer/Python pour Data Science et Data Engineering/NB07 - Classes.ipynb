# Classes

Les classes en Python permettent de créer des objets en définissant leurs attributs et méthodes. Elles servent à structurer le code selon le paradigme orienté objet, facilitant la réutilisation et l'organisation. Une classe se définit avec le mot-clé `class`, et l'initialisation des objets se fait via la méthode spéciale `__init__`. Les objets créés à partir d'une classe sont appelés instances.

Pour défnir une classe on écrit : 
```python
class ClassName():
  <code block>
```

## Exemple de classe en Python

```python
class Voiture:
    def __init__(self, marque, modele):
        self.marque = marque
        self.modele = modele

    def afficher_info(self):
        print(f"Voiture : {self.marque} {self.modele}")
```

## Création d'une instance
```python
ma_voiture = Voiture("Toyota", "Corolla")
ma_voiture.afficher_info()  # Affiche : Voiture : Toyota Corolla
```

````python
class Dog():
    pass
````

L'**instanciation** d'une classe consiste à créer un objet à partir de cette classe. On utilise le nom de la classe suivi de parenthèses et des éventuels arguments attendus par la méthode `__init__`. Par exemple :  
```python
mon_objet = MaClasse(arg1, arg2)

Ici, `mon_objet` est une instance de la classe `MaClasse`.
```

````python
my_dog = Dog()
print(type(my_dog))
````

## Réutilisation du code avec les méthodes

La réutilisation du code consiste à écrire des fonctions ou des méthodes qui peuvent être appelées plusieurs fois, évitant ainsi la duplication. Dans une classe, les **méthodes** sont des fonctions définies à l'intérieur de la classe et associées à ses objets. Elles permettent d'effectuer des actions ou de manipuler les attributs de l'objet.

Par exemple, la méthode `afficher_info` dans la classe `Voiture` peut être utilisée par toutes les instances de cette classe, ce qui évite d'avoir à réécrire le même code pour chaque objet :

```python
class Voiture:
    def __init__(self, marque, modele):
        self.marque = marque
        self.modele = modele

    def afficher_info(self):
        print(f"Voiture : {self.marque} {self.modele}")

# Réutilisation de la méthode pour différentes instances
voiture1 = Voiture("Renault", "Clio")
voiture2 = Voiture("Peugeot", "208")

voiture1.afficher_info()  # Affiche : Voiture : Renault Clio
voiture2.afficher_info()  # Affiche : Voiture : Peugeot 208
```


Cela rend le code plus propre, plus facile à maintenir et à étendre.

Appliquons cette logique à la classe `Dog()` 

````python
class Dog():

    def get_name(self, name):
        return f"name : {name}"
````

````python
my_dog = Dog()
print(my_dog.get_name("Tommy"))
````

### Qu'est-ce que `self` dans une classe Python ?

Dans une classe Python, `self` est une référence à l'instance courante de la classe. Il permet d'accéder aux attributs et méthodes de l'objet créé à partir de la classe. On le retrouve comme premier paramètre de toutes les méthodes d'instance, mais il n'est pas à renseigner lors de l'appel de la méthode.

Exemple :
```python
class Chien:
    def __init__(self, nom):
        self.nom = nom  # 'self.nom' est un attribut de l'instance

    def aboyer(self):
        print(f"{self.nom} aboie !")

mon_chien = Chien("Rex")
mon_chien.aboyer()  # Affiche : Rex aboie !
```
Ici, `self.nom` fait référence à l'attribut `nom` de l'objet `mon_chien`.

````python
class DogWithSelf():

    def print_self(self):
        print(self)

my_dog = DogWithSelf()

print(my_dog)
my_dog.print_self()
````

## Data Caching et Attributs

### Data Caching

Le **data caching** (mise en cache des données) est une technique d'optimisation des performances qui consiste à stocker temporairement des données fréquemment utilisées en mémoire ou sur disque. Sur des plateformes comme Databricks, cela permet d'accélérer l'accès aux données lors de traitements répétitifs, en évitant de relire les mêmes données depuis la source à chaque exécution. Par exemple, Spark propose des méthodes comme `.cache()` ou `.persist()` pour conserver les DataFrames en mémoire ou sur disque, réduisant ainsi le temps de calcul lors d'opérations multiples sur les mêmes données.

### Attributs

En programmation orientée objet, un **attribut** est une variable associée à une instance d'une classe (un objet). Les attributs permettent de stocker l'état ou les caractéristiques d'un objet. Par exemple, dans la classe suivante :

```python
class DogWithAttributes():
    def __init__(self, name, color):
        self.name = name  # attribut
        self.color = color  # attribut
```


Ici, `name` et `color` sont des attributs de chaque objet `DogWithAttributes`. On y accède via `self.attribut` à l'intérieur de la classe, et via `objet.attribut` à l'extérieur.

````python
class DogWithAttributes():
    def __init__(self, name, color):
        self.name = name  # attribut
        self.color = color  # attribut

my_dog = DogWithAttributes("Tommy", "brown")
print(my_dog.name)
print(my_dog.color)
````

Dans une méthode on peut accéder à un attribut en utilisant `self.attribut`

````python
class DogWithAttributesAndMethodDogWithAttributesAndMethod():
    def __init__(self, name, color):
        self.name = name  # attribut
        self.color = color  # attribut

    def return_name(self):
        return self.name
    
    def return_color(self):
        return self.color

my_dog = DogWithAttributesAndMethod("Tommy", "brown")

my_dog.return_name()
my_dog.return_color()
````

On peut rajouter une fonctionnalité par exemple modification du nom du chien

````python
class DogWithAttributesAndMethod():
    def __init__(self, name, color):
        self.name = name  # attribut
        self.color = color  # attribut

    def return_name(self):
        return self.name
    
    def return_color(self):
        return self.color
    
    def update_name(self, new_name):
        self.name = new_name

my_dog = DogWithAttributesAndMethod("Tommy", "brown")

print(f"name before update: {my_dog.return_name()}")
my_dog.update_name("Rex")
print(f"name after update: {my_dog.return_name()}")
````
