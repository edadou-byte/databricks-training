# Erreurs et exceptions en Python

En Python, les erreurs et exceptions sont des événements qui interrompent le déroulement normal d’un programme. Elles se divisent en deux grandes catégories :

- **Erreurs de syntaxe** : détectées lors de l’analyse du code (ex : parenthèse manquante).
- **Exceptions** : erreurs détectées pendant l’exécution (ex : division par zéro, accès à une clé inexistante dans un dictionnaire).

## Gestion des exceptions

On utilise les blocs `try`, `except`, `else` et `finally` pour gérer les exceptions :

```python
try:
    # Code susceptible de provoquer une exception
    resultat = 10 / 0
except ZeroDivisionError:
    print("Division par zéro !")
except Exception as e:
    print(f"Autre erreur : {e}")
else:
    print("Aucune exception n’a été levée.")
finally:
    print("Bloc finally exécuté dans tous les cas.")
```

## Exceptions courantes

- `ZeroDivisionError` : division par zéro
- `TypeError` : opération sur des types incompatibles
- `ValueError` : valeur inappropriée
- `KeyError` : clé absente dans un dictionnaire
- `IndexError` : index hors limites dans une liste

## Lever une exception

On peut lever une exception avec `raise` :

```python
raise ValueError("Message d’erreur personnalisé")
```

### Les exceptions en Python

Les **exceptions** en Python sont des erreurs détectées lors de l’exécution du programme. Elles interrompent le déroulement normal du code lorsqu’un événement inattendu survient (ex : division par zéro, accès à une clé inexistante dans un dictionnaire). Pour gérer ces situations, on utilise les blocs `try`, `except`, `else` et `finally`. Cela permet d’anticiper et de traiter proprement les erreurs, évitant ainsi l’arrêt brutal du programme.

````python
# exemple d'exception
1 / 0
````

#### Gestion des exceptions en Python

La gestion des exceptions permet de contrôler les erreurs qui surviennent lors de l’exécution d’un programme. On utilise les blocs `try`, `except`, `else` et `finally` pour intercepter et traiter ces erreurs sans interrompre brutalement le programme. Cela permet d’afficher des messages d’erreur personnalisés, d’effectuer des actions correctives ou de garantir l’exécution de certaines instructions, même en cas d’erreur. Lever une exception avec `raise` permet également de signaler explicitement une situation anormale.

**Exemple** : 
```python
try:
    code block
except:
    code block
```

Reprenons l'exemple précédent avec la division par 0

````python
try:
  1 / 0
except:
  print("On ne peut pas diviser par 0")
````

Dans ce cas cela va afficher "On ne peut pas diviser par 0" peut importe l'exception.

Si on veut gérer qu'un (ou plusieurs types) d'excéptions on écrit le nom de l'excécption aprs le `except`  :

````python
try:
    1 / 0
except ZeroDivisionError:
  print("On ne peut pas diviser par 0")
````

Qu'est ce qui se passe quand on à une autre excéption que celle renseigné ?

````python
try:
    print(variable)
except ZeroDivisionError:
  print("On ne peut pas diviser par 0")
````

Pour gérer plusieurs excéptions il faut écrire les noms entre parenthèse après le `except` : 

```python
try:
  code block
except (ExceptionName1, ExceptionName2):
  code block
```

````python
try:
    print(variable)
except (ZeroDivisionError, NameError):
  print("Exception Handled")
````

### Les AssertionError en Python

Une `AssertionError` est une exception levée lorsqu'une instruction `assert` échoue. L'instruction `assert` permet de vérifier qu'une condition est vraie pendant l'exécution du programme. Si la condition est fausse, Python lève automatiquement une `AssertionError` et interrompt le programme, sauf si l'exception est gérée.

**Exemple :**
```python
x = 5
assert x > 10, "x doit être supérieur à 10"
```

Ici, comme `x` n'est pas supérieur à 10, une `AssertionError` sera levée avec le message "x doit être supérieur à 10".

Les assertions sont souvent utilisées pour le débogage ou pour vérifier des hypothèses dans le code.

````python
assert 1 == 1
assert 1 == 2, "C'est faut"
````
