# Instructions conditionnelles et de contrôle

## Conditions `if` en Python

En Python, les instructions conditionnelles permettent d'exécuter du code selon certaines conditions.

- `if` : teste une condition.
- `elif` : teste une autre condition si la précédente est fausse.
- `else` : exécute un bloc si aucune condition précédente n'est vraie.

**Exemple :**
```python

x = 10

if x > 0:
    print("x est positif")

elif x == 0:
    print("x est nul")

else:
    print("x est négatif")

```

````python
if True:
    print("True")
else:
    print("False")
````

### Importance de l'indentation dans les instructions `if`

En Python, l'indentation est essentielle pour définir les blocs de code à exécuter selon la condition. Sans indentation correcte, Python ne sait pas quelles instructions appartiennent au `if`, `elif` ou `else`, ce qui provoque des erreurs.

**Exemple correct :**
```python

x = 5
if x > 0:
    print("x est positif")

```


**Exemple incorrect (provoque une erreur) :**
```python
  
x = 5
if x > 0:
print("x est positif")

```


L'indentation permet donc à Python de structurer le code et d'assurer son bon fonctionnement.

## Les opérateurs en Python

Les opérateurs permettent de réaliser des opérations sur des variables et des valeurs. Voici les principaux types d'opérateurs en Python :

- **Opérateurs arithmétiques** : `+`, `-`, `*`, `/`, `//`, `%`, `**`
  - Exemple : `a + b`, `a * b`, `a ** 2`

- **Opérateurs de comparaison** : `==`, `!=`, `<`, `>`, `<=`, `>=`
  - Exemple : `a == b`, `x >= 10`

- **Opérateurs logiques** : `and`, `or`, `not`
  - Exemple : `(a > 0) and (b < 5)`

- **Opérateurs d'affectation** : `=`, `+=`, `-=`, `*=`, `/=`
  - Exemple : `x += 1` (équivalent à `x = x + 1`)

- **Opérateurs d'appartenance** : `in`, `not in`
  - Exemple : `'a' in 'python'`

- **Opérateurs d'identité** : `is`, `is not`
  - Exemple : `a is b`

**Exemple d'utilisation :**
```python
a = 5
b = 2

print(a + b)      # Addition
print(a > b)      # Comparaison
print(a != b)     # Différent
print((a > 0) and (b < 5))  # Logique
```

````python
1 > 3
````

````python
print(1 == 1)
print(1.5 != 2.5)
print("abc" == "xyz")
print(True == True)
````

Utilisons maintenant des operateurs dans un if-statement

````python
lunch_price = 18

if lunch_price <= 15:
    print("Achetons le dejeuner")
else:
    print("C'est trop cher, on mange chez nous")
````

On peut mettre d'autre bloc if-else dans une codition

````python
if lunch_price <= 15:
    print("Achetons le dejeuner")
else:
    if lunch_price <= 25:
        print("Est ce que ça vaut le coup ?")
    else:
        print("C'est trop cher, on mange chez nous")
````
