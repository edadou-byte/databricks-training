# Exploration de l'architecture de Spark dans Databricks




L’architecture de Spark repose sur un modèle distribué permettant le traitement rapide de grandes quantités de données. Voici les principaux composants :

* **Driver** : Programme principal qui orchestre l’exécution des tâches, gère le contexte Spark et distribue le travail aux exécutors.
* **Executors** : Processus sur les nœuds du cluster qui exécutent les tâches, stockent les données en mémoire et renvoient les résultats au driver.
* **Cluster Manager** : Système (YARN, Mesos, Kubernetes ou Standalone) qui gère les ressources du cluster et l’allocation des exécutors.
* **RDD/DataFrame** : Structures de données distribuées permettant le calcul parallèle et la tolérance aux pannes.

Spark optimise l’exécution grâce à la gestion intelligente des tâches, la persistance en mémoire et la scalabilité sur plusieurs machines.

## SparkSession et SparkContext

**SparkSession** : Point d’entrée principal pour travailler avec Spark. Il permet de créer des DataFrames, lire des données, configurer Spark et accéder à toutes les fonctionnalités du framework.

**SparkContext** : Ancien point d’entrée utilisé pour interagir avec le cluster Spark, créer des RDD et gérer la configuration. Avec l’arrivée de SparkSession, SparkContext est désormais accessible via `spark.sparkContext`.

Dans databricks la SparkSession est créé par défaut et accessible directement avec `spark`


### Cluster Spark dans Databricks

**All-purpose compute Spark** dans Databricks désigne des clusters conçus pour l’exécution interactive de notebooks, le développement, l’exploration de données et la collaboration entre utilisateurs. Ces clusters sont adaptés aux tâches ad hoc, à l’analyse exploratoire et au prototypage, contrairement aux clusters dédiés aux jobs automatisés.

La **vue Compute** dans l’administration Databricks permet de :
* Créer, configurer et surveiller les clusters (taille, type de machine, runtime, etc.)
* Gérer les accès et permissions des utilisateurs
* Suivre l’utilisation des ressources, l’état des clusters et les logs d’activité
* Arrêter, redémarrer ou supprimer des clusters

Cette vue facilite la gestion centralisée et sécurisée des environnements de calcul pour tous les utilisateurs de la plateforme.

---

### Comparaison : All-purpose Compute vs Serverless dans Databricks

| Caractéristique         | All-purpose Compute                | Serverless Compute                |
|------------------------|------------------------------------|-----------------------------------|
| **Type de cluster**    | Cluster dédié, configurable        | Cluster géré automatiquement      |
| **Gestion des ressources** | Manuelle (taille, runtime, etc.) | Automatique, sans gestion directe |
| **Accès SparkContext** | Oui                                | Non                               |
| **Cas d’usage**        | Notebooks interactifs, prototypage | Exécution rapide, jobs éphémères  |
| **Scalabilité**        | Ajustable par l’utilisateur        | Scalabilité automatique           |
| **Administration**     | Contrôle avancé, gestion fine      | Simplifiée, moins d’options       |
| **Sécurité**           | Paramétrable                       | Intégrée, gérée par Databricks    |

**Résumé** :  
Les clusters *all-purpose* offrent plus de contrôle et de flexibilité pour le développement collaboratif, tandis que le *serverless* simplifie l’utilisation et l’administration pour des tâches rapides et automatisées.
