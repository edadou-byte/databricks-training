# Gouvernance de données et Unity Catalog

## Gouvernance de données dans Databricks et Unity Catalog

La gouvernance de données désigne l'ensemble des processus, politiques et technologies permettant de gérer, protéger et valoriser les données d'une organisation.

### Gouvernance de données dans Databricks
Databricks propose des outils pour :
- Centraliser et sécuriser les données sur le cloud.
- Contrôler l'accès aux données via des permissions sur les clusters, notebooks et fichiers.
- Assurer la traçabilité et l'audit des accès et modifications.

### Unity Catalog
Unity Catalog est la solution de gouvernance de données unifiée de Databricks :
- **Gestion centralisée des accès** : Définit des permissions fines sur les bases, tables, vues et colonnes.
- **Catalogage des données** : Organise les données en catalogues, schémas et tables pour une meilleure visibilité.
- **Traçabilité et audit** : Enregistre toutes les opérations sur les données pour répondre aux exigences de conformité.
- **Sécurité avancée** : Prend en charge le contrôle d'accès basé sur les rôles (RBAC) et l'intégration avec les identités d'entreprise.

Unity Catalog facilite ainsi la gestion sécurisée, conforme et collaborative des données sur Databricks.

## Tags et Policies dans Unity Catalog

### Tags
Les **tags** (ou étiquettes) permettent d’ajouter des métadonnées aux objets de données (tables, colonnes, schémas, etc.). Ils servent à :
- Classifier les données (ex : `confidentiel`, `PII`, `financier`).
- Automatiser l’application de règles de sécurité ou de conformité.
- Faciliter la recherche et la gestion des données sensibles.

### Policies
Les **policies** (politiques) définissent les règles d’accès et d’utilisation des données. Elles peuvent être appliquées dynamiquement en fonction des tags :
- Restreindre l’accès à certaines données selon leur classification.
- Appliquer des masques ou des règles de redaction sur les colonnes sensibles.
- Automatiser la conformité (ex : RGPD) en liant des politiques à des tags spécifiques.

**Exemple d’utilisation combinée :**
1. Taguer une colonne contenant des données personnelles avec `PII`.
2. Définir une policy qui masque automatiquement toutes les colonnes taguées `PII` pour les utilisateurs non autorisés.

Cette approche permet une gouvernance fine, centralisée et automatisée de la sécurité des données dans Databricks.

## B. Créer un groupe de sécurité sur Databricks

Pour créer un groupe de sécurité dans Databricks :

1. **Accédez à l’interface d’administration :**
   - Cliquez sur l’icône de votre profil (en haut à droite) puis sur **Admin Console**.

2. **Ouvrez l’onglet "Groups" :**
   - Sélectionnez **Groups** dans le menu latéral.

3. **Créez un nouveau groupe :**
   - Cliquez sur **Create Group**.
   - Donnez un nom au groupe (ex : `data_scientists`).
   - Cliquez sur **Create**.

4. **Ajoutez des membres au groupe :**
   - Sélectionnez le groupe créé.
   - Cliquez sur **Add Members** et choisissez les utilisateurs ou groupes à ajouter.

5. **Attribuez des permissions au groupe** selon vos besoins (accès à des clusters, tables, etc.).

> **Remarque :** Vous pouvez aussi gérer les groupes via SCIM ou l’API Databricks pour automatiser la gestion des accès.
