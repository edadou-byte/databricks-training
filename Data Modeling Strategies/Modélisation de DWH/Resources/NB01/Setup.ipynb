```python
username = spark.sql("SELECT current_user()").collect()[0][0].split("@")[0]
catalog_name = f"demo_{username}"

print(f"nom du catalogue : {catalog_name}")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name}")
```

```python
schema_names = ['bronze', 'silver', 'gold']

for schema_name in schema_names:
    print(f"création du schéma : {schema_name}")
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_name}")
```

```python
tables = spark.sql("SHOW TABLES IN samples.tpch")

df_size = tables.count()

table_names = [row.tableName for row in tables.collect()]

for table_name in table_names:
    print(f"création de la table : tpch_{table_name} dans le schéma bronze")
    spark.sql(f"""
    CREATE TABLE IF NOT EXISTS {catalog_name}.bronze.tpch_{table_name} AS
    SELECT * FROM samples.tpch.{table_name}
    """)
```
