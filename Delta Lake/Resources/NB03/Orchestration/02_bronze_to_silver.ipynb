```python
%sql
USE CATALOG demo_orchestration;
CREATE SCHEMA IF NOT EXISTS silver;
USE SCHEMA silver;
```

```python
%sql
create table if not exists communes_enrichies_silver
(
  id_commune STRING,
  nom_commune STRING,
  departement STRING,
  region STRING,
  CONSTRAINT pk_communes_enrichies PRIMARY KEY(id_commune)
)
```

```python
%sql
    
insert into communes_enrichies_silver
select 
  cb.COM as id_commune,
  cb.NCCENR as nom_commune,
  db.NCCENR as departement,
  rb.NCCENR as region
from 
  bronze.commune_bronze as cb
JOIN
  bronze.departement_bronze as db
  ON cb.DEP = db.DEP
JOIN
  bronze.region_bronze as rb
  ON cb.REG = rb.REG;
```

```python
%sql
create table if not exists crime_stats_silver
(
  id_commune STRING,
  annee int,
  indicateur STRING,
  unite_de_compte STRING,
  nombre int,
  CONSTRAINT pk_crime_stats PRIMARY KEY(id_commune, annee, indicateur)
)
```

```python
%sql
    
insert into crime_stats_silver
select 
  CODGEO_2025 as id_commune,
  annee as annee,
  indicateur as indicateur,
  unite_de_compte as unite_de_compte,
  nombre as nombre
FROM bronze.crime_stats_bronze
where est_diffuse = 'diff'
  
```
