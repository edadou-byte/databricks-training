```python
catalog_name = "demo_orchestration"

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name}")

# Définition du schéma à utiliser
schema = "bronze"
volume = "dld_demo"

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema}")

# On renseigne les catalogue / schéma à utiliser
spark.sql(f"USE CATALOG {catalog_name}")
spark.sql(f"USE {schema}")
```

```python
%sql
create table if not exists crime_stats_bronze using delta;
    
COPY INTO crime_stats_bronze
FROM (
  SELECT 
    current_timestamp() process_time,
    _metadata.file_name source_file,
    *
    FROM '/Volumes/demo_orchestration/bronze/dld_demo/orchestration/donnee-comm-data.gouv-parquet-2024-geographie2025-produit-le2025-06-04.parquet'
)
FILEFORMAT = PARQUET
COPY_OPTIONS ('mergeSchema' = 'true', 'header' = 'true');
```
