```python
%sql
USE CATALOG demo_orchestration;
CREATE SCHEMA IF NOT EXISTS gold;
USE SCHEMA gold;
```

```python
%sql
CREATE OR REPLACE VIEW crime_stats_gold
AS
SELECT 
  csc.indicateur,
  csc.unite_de_compte,
  csc.nombre,
  csc.annee,
  c.nom_commune,
  c.departement,
  c.region
FROM silver.crime_stats_silver as csc
INNER JOIN silver.communes_enrichies_silver as c
ON csc.id_commune = c.id_commune
ORDER BY nombre DESC;
```

```python
%sql
select * from crime_stats_gold;
```
