# Orchestration

## Orchestration dans Databricks

L'orchestration dans Databricks consiste à automatiser et gérer l'exécution de workflows de données, tels que des notebooks, scripts, ou jobs, afin de garantir le bon déroulement des pipelines de traitement. Databricks propose des outils comme les **Jobs** pour planifier, déclencher, et surveiller des tâches, avec des dépendances, des notifications, et des relances en cas d'échec.

## Orchestration des Lakeflow Jobs

Lakeflow est une solution d'orchestration native dans Databricks pour gérer des pipelines de données complexes. Les Lakeflow Jobs permettent de définir des étapes (tasks) avec des dépendances, d'automatiser l'ingestion, la transformation et l'export des données, et d'assurer la fiabilité du pipeline grâce à la gestion des erreurs, la reprise automatique et la surveillance centralisée. Lakeflow facilite la création de workflows modulaires et reproductibles pour les architectures lakehouse.
